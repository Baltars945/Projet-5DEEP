## 1. Analyse exploratoire

Avant de commencer on analyse le dataset pour détecter d'éventuels déséquilibres.

- **Description générale** : Nombre d'images, format des images, répartition entre train / validation et test.
- **Répartition des classes** : un gros déséquilibre pourrait pousser le modèle à mettre en avant certaines classes.
- **Des exemples d'images** : triées par classes.
- **dimensions des images** : vérifier si toutes les images ont la même taille.
- **Les couleurs** : comparer les couleurs moyennes de chaque classe.
- **Graphiques** : enregistrés dans le dossier "figures/".

## 2. Description général du dataset

On utilise le dataset **Intel Image Classification**, disponible sur **Hugging Face**. Ce dataset regroupe des photographies de paysages réparties en **6 classes** :

- **Buildings**
- **Forest**
- **Glacier**
- **Mountain**
- **Sea**
- **Street**

le dataset est fourni avec un jeu d'entrainement et jeu de test
Découpage du jeu d'entrainement en deux partie :
- **train (85%)** : utilisée pour apprendre les poids du modèle.
- **validation (15%)** : utilisée pour suivre l'entrainement et déclencher un arrêt anticipé

Chaque classe est représentée dans les mêmes proportions dans le train et dans la validation, tout en étant **reproductible** grâce à une graine aléatoire fixe (**SEED**).

Le jeu de test n'est utilisé que pour mesurer la performance finale du modèle sur des images qu'il na jamais vue.

## 3. Format et prétraitement des images

Dans leur forme de base, les images sont stockée au format **.JPEG en couleur** de tailles, pour la plpart, de **150x150 pixels**.

Chaque images subit un prétraitement : 
- **Conversion en RGB** : pour garantir 3 canaux de couleur même si une image était en couleur foncé.
- **Redimensionnement en 128x128 pixels** : pour garantir que toutes les images ont bien la même tailles.
- **Conversion en tenseur PyTorch** : (canaux, hauteur, largeur)

La taille définit de **128x128** permet de réduire le temps de calcul par rapport à du **150x150** tout en conservant une bonne résolution d'image, elle est aussi divisibles par 16, ce qui est nécessaire pour les 4 couches de **max pooling** successives du CNN.

## 4. Répartition des classes

On compte le nombre d'images pour chaque classes dans les trois sous-ensemble (train, validation, test), en valeur absolue et en pourcentage.

**importance** : si une classe est beaucoup plus répresentée que les autres, le modèle aura tendance à le prédire plus souvent, ce qui apour cause de provoquer un déséquilibre entre les classes et fausser les résultat et les performances du models.

**Comment le mesuré** :
- Calcul du ration entre la classe la **plus représenter** et la classe la **moins rprésenter**.
- Un ration proche de **1** indique des classes équilibrées.
- un ratio élevé indique qu'il faut corriger le déséquilibre.

On vérifie également que les proportions entre train, validation et test soit similaires pour garantir que les trois sous-ensemble sont représentatifs du même problème.

## 5. Exemple d'images par classe

On affiche plusieurs images aléatoire pour chaque classe, afin d'observer le contenue du dataset.

Cette observation permet :
- Vérifier que les images correspondent bien à leur classe.
- apprendre au model à connaitre plusieur angle de vue pour une même classe.
- Identifier des classes qui peuvent être visuellement proches :
    - **Street / Buildings**

Les classes qui se ressemnle sont à surveillé pour éviter toute erreur.

## 6. Dimensions des images

On repére la taille d'origine d'une image (**largeur * hauteur**), puis on compare le nombre d'image avec une taille standard et le nombre d'image avec une taille différente.

**Pourquoi ?**
- Un réseaux de neurones convolutif attend des entrées de tailles fixe.
- Des images de tailles différente doivent être redimensionné pour éviter tout conflit ou déformation.

Le redimensionnement en **128x128** lors du prétraitement garantit que l'entrée reste exactement identique, et comme la part des images de tailles non standard sont minotaire l'impact sur le models reste minimes.

## 7. Statistique de couleur

On calcule pour chaque classes l'**intensité moyenne** des troix canaux de couleur primaire (**Rouge, Vert, Bleu**), a partir des échantillon d'imagee.

**Pourquoi ?**
- Repérer si une couleur est présente de maniére imposante dans une classe (**sea : Bleu**) ou (**Forest : Vert**) pour savoir si la couleur consitue une information utile que le CNN pourra exploiter.

## 8. Bilan

Cette analse permet de dégager plusieurs points qui orientent la suite du projet :
- **Equilibre des classes** : la répartition des classes détermine s'il est nécessaire de mettre en place une stratégie de rééquilibrage.
- **Taille des images** : le redimensionnement en 128x128 uniforme les entrées du réseau et réduit le coût de calcul par rapport au standard de 150x150.
- **Classe proches** : Les paires glacier/mountain et buildings / street représentent un risque de confusion, il faut bien les surveillé lors de l'évaluation.
- **Couleur** : Les différences de couleur entre classes constituent une information exploitable par le modèle.
- **Découpage des données** : le découpage stratifié et reproductible garantit une évaluation fiable et comparable entre les différents modèles testés.
